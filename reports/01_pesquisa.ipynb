## 4. Decisões e Justificação

| Decisão | Alternativas | Justificação |
|---|---|---|
| **Execução 100% Local (Antigravity)** | Treino em cloud externa (Databricks) | O utilizador definiu execução e treino exclusivamente locais, eliminando credenciais remotas, custos de instâncias e dependências externas, garantindo total privacidade e segurança. |
| **Combinação Sinérgica Enriquecida: OMIE + ENTSO-E + Open-Meteo + Mercados Financeiros** | Utilizar apenas OMIE ou apenas ENTSO-E isoladamente | Em vez de tratar OMIE e ENTSO-E como opções mutuamente exclusivas, a sua combinação coordenada é a abordagem ótima: o **OMIE** fornece o preço oficial do leilão e volumes de equilíbrio marginal, enquanto a **ENTSO-E** disponibiliza as grandezas físicas fundamentais da rede (despacho por tecnologia solar/eólica/hídrica/gás, carga real e fluxos de interligação com Espanha). O cruzamento de ambos com a meteorologia (**Open-Meteo**) e os sinais de mercado de capitais enriquece dramaticamente o dataset com features de alta relevância preditiva. |
| **Incorporação de Futuros de Gás Natural (TTF) e Carbono (KRBN) com Forward-Fill Causal** | Excluir matérias-primas devido a feriados e fins de semana | Na microeconomia marginalista do MIBEL, as centrais de ciclo combinado a gás (CCGT) fixam o preço marginal na maioria das horas de ponta noturna. O seu custo é determinado pelo *Clean Spark Spread* (gás TTF + licenças de emissão de CO₂). A descontinuidade temporal em fins de semana e feriados bancários é rigorosamente mitigada aplicando *forward-fill* causal do último fecho oficial conhecido antes de $D-1$ 12:00 CET, assegurando zero *lookahead leakage*. |
| **Inclusão de Ações de Utilities Ibéricas e ETFs Setoriais** | Basear o modelo unicamente em meteorologia e preços passados | As cotações de empresas elétricas (EDP, EDPR, Iberdrola, REN) e ETFs como ICLN/XLE agregam as expectativas de mercado sobre margens de produção, riscos hidrológicos e fluxos de investimento verde, oferecendo sinais exógenos valiosos com disponibilidade pública e diária. |
| **Resolução Horária com Média Aritmética para 15 min ($\frac{P_{15} + P_{30} + P_{45} + P_{60}}{4}$)** | Interpolação spline ou agregação ponderada | A média aritmética simples dos 4 períodos de 15 minutos ($\frac{P_{15} + P_{30} + P_{45} + P_{60}}{4}$) é a convenção regulatória europeia oficial (SDAC / MTU15) para apuramento do preço médio horário equivalente pós 01/10/2025. |
| **Enquadramento de ML: Regressão Contínua do Preço Horário (€/MWh)** | Classificação Binária de Picos ou Deteção de Anomalias | A regressão contínua fornece o valor económico exato do megawatt-hora, indispensável para otimização de despacho térmico, arbitragem com baterias, estratégias de contratação e cálculo de custos de desvio. |

## 6. Problemas, Riscos e Limitações Encontrados

1. **Sinergia e Combinação OMIE + ENTSO-E (Harmonização e Robustez):**
   - *Vantagem e Complexidade:* A combinação do OMIE (cotações de leilão financeiro) com a ENTSO-E (despacho físico real e previsto por tecnologia) cria uma sinergia ímpar. O desafio reside na harmonização de calendários e na agregação estrita ao fuso `Europe/Lisbon` sem introduzir desfasamentos temporais.
   - *Mitigação:* O Agente 2 agregará os dados através da chave temporal comum `timestamp_utc`, mantendo imutáveis as respostas brutas das APIs em `data/raw/`.

2. **Mercados de Futuros de Commodities (Gás Natural TTF e Carbono KRBN):**
   - *Limitação Operacional:* Os mercados financeiros de matérias-primas não negoceiam aos fins de semana e encerram em feriados bancários internacionais, gerando descontinuidades temporais em relação ao mercado elétrico (que opera 365 dias/ano, 24 horas/dia).
   - *Mitigação Analítica:* Aplicação de *forward-fill* estritamente temporal no pré-processamento. Para os sábados, domingos e feriados, utiliza-se o último preço de fecho oficial conhecido antes de $D-1$ 12:00 CET. Esta técnica simula exatamente a informação real de que os operadores de mercado dispõem ao submeterem as suas ofertas no leilão diário, garantindo **zero data leakage**.

3. **Ações de Utilities Ibéricas e ETFs Globais (Diferencial de Horário de Negociação):**
   - *Limitação:* As bolsas de valores (Euronext Lisboa, BME Madrid, NYSE) encerram as sessões às 16:30 ou 17:30 CET, enquanto o leilão diário MIBEL fecha às 12:00 CET de $D-1$.
   - *Mitigação Anti-Leakage:* No cálculo das features derivadas de ações e ETFs, utilizam-se estritamente as cotações de fecho de $D-2$ ou o preço de abertura (*Open*) da manhã de $D-1$, prevenindo a incorporação de preços de fecho da tarde de $D-1$ que violariam a causalidade temporal.

4. **Execução 100% Local (Sem Dependência de Cloud Externa):**
   - *Decisão Técnica:* O projeto corre integralmente em ambiente local Python, sem necessidade de chaves pagas, instâncias externas ou transmissão de dados para fora do ambiente de desenvolvimento do utilizador.
